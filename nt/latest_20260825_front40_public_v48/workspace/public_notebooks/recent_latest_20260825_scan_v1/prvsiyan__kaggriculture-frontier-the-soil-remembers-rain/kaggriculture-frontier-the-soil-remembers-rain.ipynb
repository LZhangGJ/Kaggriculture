# Kaggriculture Frontier | The Soil Remembers Rain

**V139: public-state Moon/Soil loss router.** V139 keeps the stronger V135 Moon policy by default and switches to a repaired Soil V26-H continuation only for early public farm states that repeatedly defeated Moon and do not collide with the current top panel.


````python
import base64, hashlib, importlib.metadata, tarfile, warnings, zlib
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')
CANDIDATE_SHA = 'ff871d2fbf2b1123668045062677eea7f0239fc95b02d15fedce35c2c1033fd1'
EVIDENCE_ENGINE = '1.32.7'
RUNTIME_ENGINE = importlib.metadata.version('kaggle-environments')
assert RUNTIME_ENGINE == EVIDENCE_ENGINE
print('Candidate SHA:', CANDIDATE_SHA)
print('Runtime/evidence engine:', RUNTIME_ENGINE)
````

## Strategy

V139 runs the complete V135 observation-driven policy unless the rival's first resolved public farm state matches one of five validated opening families: two hands with $167; four hands with one of $4, $7, $71, $76, or $2,013 and the observed pasture shape; five hands with $3 and no pasture; or five hands with $2,988 and one pasture. Those classes switch to embedded Soil V26-H after a dependency repair that makes Soil compatible with V135's shared first action.

The selector uses only current public rival farm state—hand count, hires, pasture count, and money. It does not read identity, team, rating, episode seed, reward, future shops, or private rival state. Five-hand/$19 is explicitly excluded because candidate head-to-head evaluation proved that it collides with top MiMi even though replay-only inspection showed $20.


````python
validation = pd.DataFrame([
    ['Current top-10 exact routes', 100, 70, 30, 11506.8],
    ['All 17 Moon public losses', 34, 20, 14, 1774.7],
    ['V135 on same loss panel', 34, 4, 30, -3267.4],
], columns=['panel','games','wins','losses','mean_margin'])
display(validation)
ax=validation.set_index('panel')[['wins','losses']].plot.bar(
    color=['#5D8C63','#B84A62'], figsize=(10,4), rot=10,
    title='Frozen V139 official-engine evidence')
ax.set_ylabel('games'); plt.tight_layout()
````

## Frozen selection result

Every one of Moon V135's 17 completed public losses was downloaded and converted into an execution-guarded opponent. On the original seeds in both candidate seats, V139 improved the record from 4-30 to 20-14, produced sixteen loss-to-win flips and zero win-to-loss flips, recovered ten loss episodes in both seats, raised mean margin by 5,042.1, and improved the worst margin from -29,514 to -13,388.

The complete latest top-10 panel was then rebuilt from the five newest public routes for each exact leading submission. V139 was reward-, opponent-reward-, margin-, and outcome-exact to V135 on all 100 paired rows, preserving V135's 70-30 record and +11,506.8 mean margin.


````python
comparison = pd.DataFrame([
    ['Moon-loss paired rows', 34, 17, 14, 3, 16, 0, 5042.1],
    ['Current-top paired rows', 100, 0, 100, 0, 0, 0, 0.0],
], columns=['evidence','rows','improved','exact','regressed','loss_to_win','win_to_loss','mean_delta'])
display(comparison)
````

## Promotion and falsification gates

V138 originally routed the five-hand/$19 class because it covered five Moon losses. The fresh top panel falsified that selector: all ten MiMi rows entered the same class, four V135 wins became losses, and mean margin fell by 1,147.8. V138 was rejected.

V139 removes exactly that ambiguous class. It preserves the remaining ten public-loss episodes, including the $167, low-money four-hand, $2,013, $3, and $2,988 openings, while becoming exactly V135 on every current-top row. The common V92-style five-hand/$13 population that caused thirty-five Soil losses also stays on Moon.


````python
cohort = pd.DataFrame([
    ['Current top-10', '70-30', '+11,506.8', 'exactly V135'],
    ['Moon public losses', '20-14', '+1,774.7', '10/17 episodes recovered both seats'],
    ['Rejected V138 top-10', '66-34', '+10,359.0', '4 MiMi win-to-loss flips'],
], columns=['source cohort','record','mean margin','boundary'])
display(cohort)
````

## Mechanism evidence

Pure Soil is not globally deployable: thirty-five of its forty-four observed losses shared a V92-style five-hand, pasture, $13 state and averaged -9,091.7. Conversely, Soil won 32 of 34 guarded reruns against Moon's actual loss opponents while Moon won only four. V139 treats these as complementary policies rather than averaging them: Moon protects the common V92 and top-agent classes, while repaired Soil handles only non-colliding loss fingerprints.


````python
evidence = pd.DataFrame([
    ['candidate_sha256','ff871d2fbf2b1123668045062677eea7f0239fc95b02d15fedce35c2c1033fd1'],
    ['engine','1.32.7'],
    ['current_top','70-30; mean +11,506.8'],
    ['current_top_pairing','100/100 exact to V135'],
    ['moon_loss_panel','20-14; mean +1,774.7'],
    ['loss_pairing','17 improved / 14 exact / 3 still-winning regressions'],
    ['outcome_flips','16 loss-to-win / 0 win-to-loss'],
    ['episodes_recovered','10/17 in both seats'],
    ['known_boundary','two-hand/$280 and five-hand/$19 remain Moon'],
    ['competition_submission','pending this notebook version'],
], columns=['field','value'])
display(evidence)
````

## Why this composition survived

The selector is deliberately small because action-dependent market clearing makes replay-only opening signatures unsafe. Each retained class passed two tests: it recovered the corresponding deployed-Moon losses in both candidate seats, and it remained inactive across all 100 refreshed top rows. The $19 class failed the second test and was removed even though it looked excellent on the loss panel.


````python
rejections = pd.DataFrame([
    ['Pure Soil V26-H', '35 losses to common five-hand/$13 V92-style openings'],
    ['V137 raw V92 opening', '22/34 paired loss rows regressed; no outcome recovery'],
    ['V138 including five-hand/$19', 'collided with MiMi; four top wins became losses'],
    ['Two-hand/$280 switch', 'same public state hides both Soil win and severe Soil loss'],
    ['Identity/rating/seed routing', 'non-causal, unavailable, and prohibited'],
], columns=['rejected intervention','reason'])
display(rejections)
````

## Live execution proof

The exact emitted `main.py` plays complete official-engine episodes from both seats. This is the runtime and serialization gate before the tied notebook creates a public competition submission.


````python
from kaggle_environments import make

live_rows=[]
for side,seed in [(0,70110),(1,70111)]:
    agents=[str(MAIN_PATH),'starter'] if side==0 else ['starter',str(MAIN_PATH)]
    env=make('kaggriculture',configuration={'episodeSteps':720,'seed':seed},debug=False)
    env.run(agents)
    mine=env.state[side]
    assert mine.status=='DONE',(side,mine.status)
    live_rows.append({'candidate seat':side,'seed':seed,'status':mine.status,'reward':mine.reward})
display(pd.DataFrame(live_rows))
````

## Package integrity

The reproducibility bundle contains exactly one root-level `main.py`; the notebook reopens it and verifies the packed bytes. The public selector and both policy bodies are self-contained and require no internet, dataset, identity, or hidden state at match time.


````python
BUNDLE_PATH = Path("submission.tar.gz")
with tarfile.open(BUNDLE_PATH, "w:gz") as bundle:
    bundle.add(MAIN_PATH, arcname="main.py")

with tarfile.open(BUNDLE_PATH, "r:gz") as bundle:
    names = bundle.getnames()
    packed = bundle.extractfile("main.py").read()

assert names == ["main.py"]
assert hashlib.sha256(packed).hexdigest() == EXPECTED_AGENT_SHA256
print("bundle members", names)
print("bundle sha256", hashlib.sha256(BUNDLE_PATH.read_bytes()).hexdigest())
````
