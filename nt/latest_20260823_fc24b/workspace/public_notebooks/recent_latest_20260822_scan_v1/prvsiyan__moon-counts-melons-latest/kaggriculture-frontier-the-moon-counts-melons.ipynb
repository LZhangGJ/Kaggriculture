**V92: Causal public-state routing with bounded scarcity production.** This version replaces the stale public V56 bytes with the strongest frozen Moon agent from the local campaign. Its optional carrot, tomato, egg, strawberry-flow, capacity, and terminal modules activate only from observations available at decision time; inactive branches preserve the underlying route exactly.

## Strategy

V92 preserves Moon's complete current/legacy farm routes and chooses among coherent livestock/crop continuations from public shop evidence. V71 contributes a route-owned carrot relay; the tomato conversion and strawberry-flow overlays require live scarcity, visible rival supply, and route compatibility; the egg path is a complete buy/build/place cohort with strict fallbacks. A late tomato controller is admitted only when post-supply deficit remains at least 300 units. Capacity protection and the live terminal controller use observed loads, shed space, prices, yields, and remaining time.

Every branch is bounded. If its public gate fails, the exact base action object is returned. The policy reads no episode seed, opponent identity, reward, hidden future shop, or coordinate/RNG fingerprint.


````python
fit = pd.DataFrame([
    ['Deduplicated comprehensive panel', 274, 202, 72, 0, 8941.2],
    ['3165+ labeled references', 36, 30, 6, 0, 18376.4],
    ['Exact public 2913 reliability', 30, 27, 3, 0, 8791.1],
    ['Score-stratified panel', 56, 52, 4, 0, 15695.8],
    ['Current-top route panel', 42, 41, 1, 0, 16379.8],
], columns=['panel','games','wins','losses','ties','mean_margin'])
display(fit)
ax=fit.set_index('panel')[['wins','losses','ties']].plot.bar(
    color=['#c99a39','#5865A6','#9b9b9b'], figsize=(10,4), rot=13,
    title='Frozen V92 validation panels')
ax.set_ylabel('games'); plt.tight_layout()
````

## Frozen selection result

V92 was selected by row-paired, both-seat tests against exact public-source hashes. Across 274 deduplicated matches it won 202 and lost 72 with mean margin +8,941. The 36-game panel of labeled 3165+ references was 30-6. Most importantly for the original deployment gate, the exact byte-deduplicated public 2913 source was tested on 15 seeds in both seats: V92 went 27-3 with mean margin +8,791 and a 95% Wilson lower bound of 74.4%.

No later experimental branch is included here. Several attractive average-gain variants were rejected because they regressed individual held-out rows.

## Live execution proof

Two deterministic episodes exercise the exact materialized V92 agent through all 720 calls in both seats using engine 1.32.7. This is a runtime/package gate; only a completed tied-notebook competition submission produces a public score.


````python
runs=[]
for seat,seed in [(0,62010),(1,62011)]:
    agents=[str(MAIN_PATH),'starter'] if seat==0 else ['starter',str(MAIN_PATH)]
    env=make('kaggriculture',configuration={'episodeSteps':720,'seed':seed},debug=False)
    env.run(agents); mine=env.state[seat]
    assert mine.status=='DONE',(seat,mine.status)
    runs.append({'runtime_engine':RUNTIME_ENGINE_VERSION,'seat':seat,'seed':seed,'status':mine.status,'reward':mine.reward})
display(pd.DataFrame(runs))
````

## Root-only bundle

The final code cell writes one archive member, `main.py`, and proves its bytes equal the displayed candidate hash.


````python
ARCHIVE=Path('submission.tar.gz')
with tarfile.open(ARCHIVE,'w:gz') as tf:
 info=tarfile.TarInfo('main.py'); info.size=len(candidate_bytes); info.mtime=0; info.mode=0o644
 tf.addfile(info,io.BytesIO(candidate_bytes))
with tarfile.open(ARCHIVE,'r:gz') as tf:
 names=tf.getnames(); bundled=tf.extractfile('main.py').read()
assert names==['main.py'] and bundled==candidate_bytes
print('Bundle verified:',hashlib.sha256(ARCHIVE.read_bytes()).hexdigest())
````
